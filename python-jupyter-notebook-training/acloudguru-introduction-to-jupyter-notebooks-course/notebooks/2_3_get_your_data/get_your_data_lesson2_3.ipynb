```python
# read in movie CSV data and set Title as index column
df1 = pd.read_csv("IMDB-Movie_lesson_2_3.csv", index_col="Title")
```

```python
# verify data was read in - head by default reads 1st lines
df1.head()
```

```python
# read last 2 lines
df1.tail(2)
```

```python
# rename columns with lowercase and shorter names
df1.columns = ['rank','genre','desc','dir','actor','year','length','rating','votes','rev','mscore']
df1.head(2)
```

```python
# change 'index' name to 'title' (lowercase)
df1.index.name = 'title'
```

```python
# use info() to report column related info (column name, non-null count, datatype)
df1.info()
```

```python
# use describe() to report the count, mean, std deviation, min, max, etc
df1.describe()
```

```python
# use 'share' report number of columns and rows in dataset
df1.shape
```

```python
# report the number of non-null values
df1.isnull().sum()
```

```python
# replace 'rev' column NaN (not a number) values with the 'rev' mean value
df1['rev'].fillna(df1['rev'].mean(),inplace=True)
# replace 'mscore' [metascore] column NaN (not a number) values with the 'mscore' mean value
df1['mscore'].fillna(df1['mscore'].mean(),inplace=True)
```

```python
# after changs, report the number of non-null values
df1.isnull().sum()
```

```python
# drop the rank column - Note: axis=0 impacts rows, and axis=1 impacts the columns
df1.drop(['rank'],axis=1,inplace=True)
```

```python
# find mean of the movies grouped by year
df1.groupby(['year'])[['length','rating','votes','rev','mscore']].mean()
```

```python
# find mean of the movies grouped by year sorted by revenue
df1.groupby(['year'])[['length','rating','votes','rev','mscore']].mean().sort_values(by=['rev'], ascending=False)
```

```python
# sort 2012 movies by revenue
df1[df1['year'] == 2012].sort_values(by=['rev'], ascending=False)
```

```python
# find highest rated movies in the Comedy and Action genres
df1[df1['genre'].isin(['Comedy', 'Action'])].sort_values(by=['rating'], ascending=False)
```

```python
# find the best rated (in the top 99%) with the lowest revenue movies (in the bottom 25%)
df1[(df1['rating'] > df1['rating'].quantile(0.99)) & (df1['rev'] < df1['rev'].quantile(0.25))]
```
